```python
import pandas as pd
import requests
import os
```

```python
df = pd.read_csv(
    "data/observations-668356.csv",
    header=None,
    names=["image_id", "url", "species", "common_name"],
)

df.head()
print(repr(df.loc[1, "url"]))

```

```python
IMAGE_DIR = r"D:\datasets\mountain_flowers" # put your path wanted here
os.makedirs(IMAGE_DIR, exist_ok=True)
```

```python
def download_image(row):
    species_dir = os.path.join(IMAGE_DIR, row["species"].replace(" ", "_"))
    os.makedirs(species_dir, exist_ok=True)
    print (f'Downloading image {row["image_id"]} of species {row["species"]}')

    filename = f'{row["species"]}_{row["image_id"]}.jpg'
    filepath = os.path.join(species_dir, filename)

    if os.path.exists(filepath):
        return "already downloaded"

    try:
        r = requests.get(row["url"], stream=True)
        r.raise_for_status()

        with open(filepath, "wb") as f:
            for chunk in r.iter_content(chunk_size=8192):
                f.write(chunk)
            
    except Exception as e:
        return f"error: {e}"
    
print(df.iloc[[1]])
```

```python
#download images
for i in range(1,len(df)):  # change range as needed
    result = download_image(df.iloc[i])
    print(f'Image {i}/{len(df)-1}: Done')
```
